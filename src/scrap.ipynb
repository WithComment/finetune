```python
from datasets import load_dataset

sqa = load_dataset("derek-thomas/ScienceQA", split="train")
```

```python
with_image = 0
for example in sqa['image']:
  if example is not None:
    with_image += 1
    
print(f"Number of examples with images: {with_image} out of {len(sqa)}")
```
